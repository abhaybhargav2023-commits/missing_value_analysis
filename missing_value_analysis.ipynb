```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
from matplotlib.colors import ListedColormap

# -------------------------------
# 1. Calculate missing-value data
# -------------------------------
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

columns_with_missing = missing_summary[
    missing_summary["Missing Count"] > 0
].sort_values("Missing Count", ascending=True)

total_missing = int(df.isnull().sum().sum())
columns_affected = len(columns_with_missing)
total_cells = df.shape[0] * df.shape[1]
missing_percent = (total_missing / total_cells * 100) if total_cells else 0

# -------------------------------
# 2. Create dashboard
# -------------------------------
fig = plt.figure(figsize=(15, 10), facecolor="#F3F6FC")

fig.text(
    0.07, 0.94, "DATA QUALITY REPORT",
    fontsize=24, fontweight="bold", color="#172554"
)
fig.text(
    0.07, 0.90,
    "DAY 21  |  MISSING VALUE ANALYSIS  |  PANDAS",
    fontsize=11, color="#64748B"
)

# KPI card helper
def add_card(x, y, width, height, title, value, color):
    card = FancyBboxPatch(
        (x, y), width, height,
        boxstyle="round,pad=0.012,rounding_size=0.02",
        transform=fig.transFigure,
        facecolor="white",
        edgecolor="#E2E8F0",
        linewidth=1.2
    )
    fig.patches.append(card)

    fig.text(
        x + 0.025, y + height * 0.65, title,
        fontsize=10, color="#64748B"
    )
    fig.text(
        x + 0.025, y + height * 0.22, value,
        fontsize=23, fontweight="bold", color=color
    )

add_card(0.07, 0.74, 0.25, 0.11,
         "TOTAL ROWS", f"{df.shape[0]:,}", "#2563EB")
add_card(0.375, 0.74, 0.25, 0.11,
         "MISSING VALUES", f"{total_missing:,}", "#E11D48")
add_card(0.68, 0.74, 0.25, 0.11,
         "AFFECTED COLUMNS", f"{columns_affected}", "#7C3AED")

# -------------------------------
# 3. Missing values by column
# -------------------------------
ax1 = fig.add_axes([0.09, 0.39, 0.82, 0.27])
ax1.set_facecolor("white")

if not columns_with_missing.empty:
    values = columns_with_missing["Missing Count"]
    bars = ax1.barh(
        columns_with_missing.index,
        values,
        color=plt.cm.RdYlBu_r(
            np.linspace(0.15, 0.85, len(values))
        ),
        height=0.58
    )

    max_value = max(values.max(), 1)
    ax1.set_xlim(0, max_value * 1.22)

    for bar, (_, row) in zip(bars, columns_with_missing.iterrows()):
        count = int(row["Missing Count"])
        pct = row["Missing Percentage"]
        ax1.text(
            bar.get_width() + max_value * 0.02,
            bar.get_y() + bar.get_height() / 2,
            f"{count:,}  ({pct:.1f}%)",
            va="center", fontsize=10, color="#334155",
            fontweight="bold"
        )

    ax1.set_title(
        "Missing Values by Column",
        loc="left", fontsize=15, fontweight="bold",
        color="#172554", pad=15
    )
    ax1.set_xlabel("Number of missing values", color="#64748B")
    ax1.grid(axis="x", linestyle="--", alpha=0.25)
    ax1.set_axisbelow(True)
    ax1.spines[["top", "right", "left"]].set_visible(False)
    ax1.spines["bottom"].set_color("#CBD5E1")
    ax1.tick_params(axis="y", labelsize=11, length=0)
    ax1.tick_params(axis="x", colors="#64748B")
else:
    ax1.text(
        0.5, 0.5, "Great! No missing values found.",
        ha="center", va="center", fontsize=16,
        color="#059669", transform=ax1.transAxes
    )
    ax1.set_axis_off()

# -------------------------------
# 4. Missingness pattern heatmap
# -------------------------------
ax2 = fig.add_axes([0.09, 0.10, 0.82, 0.20])
ax2.set_facecolor("white")

if not columns_with_missing.empty:
    selected_cols = columns_with_missing.index.tolist()
    sample = df[selected_cols].isna().head(100).T.astype(int)

    ax2.imshow(
        sample,
        aspect="auto",
        interpolation="nearest",
        cmap=ListedColormap(["#DCEAFE", "#F43F5E"]),
        vmin=0, vmax=1
    )

    ax2.set_title(
        "Missingness Pattern  |  First 100 Records",
        loc="left", fontsize=14, fontweight="bold",
        color="#172554", pad=12
    )
    ax2.set_ylabel("Columns", color="#64748B")
    ax2.set_xlabel("Record index", color="#64748B")
    ax2.set_yticks(range(len(selected_cols)))
    ax2.set_yticklabels(selected_cols, fontsize=9)
    ax2.tick_params(axis="x", labelsize=8)
    ax2.tick_params(length=0)
else:
    ax2.text(
        0.5, 0.5, "No missing-data pattern to display.",
        ha="center", va="center", transform=ax2.transAxes
    )
    ax2.set_axis_off()

# Footer
fig.text(
    0.07, 0.045,
    f"Dataset: {df.shape[0]:,} rows × {df.shape[1]} columns"
    f"   |   Overall missing cells: {missing_percent:.2f}%",
    fontsize=10, color="#64748B"
)
fig.text(
    0.73, 0.045, "PYTHON  •  PANDAS  •  MATPLOTLIB",
    fontsize=9, color="#2563EB", fontweight="bold"
)

# Export high-resolution image
plt.savefig(
    "day21_missing_values_dashboard.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()
```
